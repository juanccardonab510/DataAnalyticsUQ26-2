1. Para junio de 2026, la tasa de desocupacion del total nacional fue 8.0%

la tasa de desempleo se calcula mediante la formula:
TD=D/PEA*100
D: cantidad de personas desempleadas
PA: poblacion economicamente activa
*En esta tabla los desempleados reales son DSI

```python
D = desempleados[desempleados['DSI']==1]['FEX_C18'].sum()
D
# cantidad de desempleados en colombia
```

```python
PEA = ft[ft['FT']==1]['FEX_C18'].sum()
PEA
# Cantidad de personas en la PEA en colombia 2134634.952160001
```

```python
TD = D/PEA*100
print(f'La tasa de desempleo en colombia para junio 2026 es de {TD} % ')
```
